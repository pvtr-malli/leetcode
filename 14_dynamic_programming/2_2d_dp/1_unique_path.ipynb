```python
class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        dp = [[1] * n for _ in range(m)]
        print(dp)

        for i in range(1, m):
            for j in range(1, n):
                dp[i][j] = dp[i-1][j] + dp[i][j-1]

        return dp[-1][-1]

Solution().uniquePaths(3, 7)


# Initialize with 1s — the entire first row and column are naturally 1 (only one path to reach any edge cell),
#  so pre-filling with 1 handles all base cases without any if check inside the loop.
```
